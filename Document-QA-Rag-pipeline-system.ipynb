```python
import gradio as gr

def ask_document(question):

    docs = vectorstore.similarity_search(
        question,
        k=3
    )

    context = "\n".join(
        [doc.page_content for doc in docs]
    )

    prompt = f"""
    Answer ONLY from the context below.

    Context:
    {context}

    Question:
    {question}
    """

    response = model.generate_content(
        prompt
    )

    return response.text
```

```python
interface = gr.Interface(
    fn=ask_document,

    inputs=gr.Textbox(
        label="Ask Question"
    ),

    outputs=gr.Textbox(
        label="Answer"
    ),

    title="Document Q&A System",

    description="""
Upload document → Ask questions → Get AI answers
"""
)

interface.launch()
```
